```python
df1.drop_duplicates(inplace=True)
```

```python
df1 = df1[['Sentiment','Text']]
```

##### Remove URLs

```python
df1['Text'] = df1['Text'].str.replace(r"http\S+|www\S+", "", regex=True)
```

##### Remoev @ mentions


```python
import re

df1["Text"] = df1["Text"].apply(lambda x: re.sub(r"@\w+", "", x))
```

##### Remove Punctuations

```python
df1["Text"] = df1["Text"].apply(lambda x: re.sub(r"[^\w\s]", "", x))
```

#### handle elongated words

```python


def remove_elongated_words(text):
    return re.sub(r'(.)\1{2,}', r'\1\1', text)

df1["Text"] = df1["Text"].apply(remove_elongated_words)
```

#### REMOVE NO_ASCII SPECIAL CHARACTER THAT CREATED DUE TO ENCOING


```python
# Remove non-ASCII special characters (Ã, Â, etc.)
df1['Text'] = df1['Text'].str.replace(r'[^\x00-\x7F]+', '', regex=True)
```

#### SAVING FILE BACK 

```python
df1.to_csv("Sentiment_train_cleaned.csv", index=False)
```
