```python
df3 = pd.read_csv("D:\\ML CODES\\IBM\\test\\socialmedia-test.csv",encoding='latin-1')
```

```python
df3.head()
```

```python
df3 = df3[['text']]
```

```python
df3.head(60)
```

```python
print("df3 duplicates:", df3.duplicated().sum())
```

```python
df3.drop_duplicates(inplace=True)
```

##### Remove URLs

```python
df3['text'] = df3['text'].str.replace(r"http\S+|www\S+", "", regex=True)
```

##### Remoev @ mentions


```python
import re

df3["text"] = df3["text"].apply(lambda x: re.sub(r"@\w+", "", x))
```

##### Remove Punctuations

```python
df3["text"] = df3["text"].apply(lambda x: re.sub(r"[^\w\s]", "", x))
```

#### handle elongated words

```python


def remove_elongated_words(text):
    return re.sub(r'(.)\1{2,}', r'\1\1', text)

df3["text"] = df3["text"].apply(remove_elongated_words)
```

```python
df3.to_csv("socialmedia_test_cleaned.csv", index=False)
```
