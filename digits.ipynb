#### Building a digit recognition model

```python
import numpy as np
import pandas as pd
import sklearn
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
import plotly.express as px
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import load_digits
digits = load_digits()
print('digits.images:' + str(digits.images.shape))
print('digits.target:' +str(digits.target.shape))
```

```python
digits.images[0]
```

```python
plt.tick_params(axis='both', which='both',bottom=False , top = False,left= False, right= False,labelbottom=False , labelleft= False)
plt.imshow(digits.images[2],cmap=plt.cm.gray_r)
```

#### cmap stands for colormap 
 it determines how the numerical pixel values in an image are mapped to colors when displaying the image.

```python
# Display the first 10 digit images (0-9)
fig, axes = plt.subplots(2, 5, figsize=(12, 5))
axes = axes.flatten()

for i in range(10):
    axes[i].imshow(digits.images[i], cmap=plt.cm.gray_r)
    axes[i].set_title(f'Digit: {digits.target[i]}')
    axes[i].axis('off')  # Hide axes

plt.tight_layout()
plt.show()
```

```python
plt.xticks([])  # Hide x-axis tick marks
plt.hist(digits.target, edgecolor='black')
```

The enumerate() function in Python is used to loop over an iterable (like a list, tuple, or string) and get both the index and the value at the same time.

```python
x_train, x_test, y_train, y_test = train_test_split(digits.data, digits.target, test_size=0.2, random_state=42)

model = LogisticRegression(max_iter=5000)
model.fit(x_train,y_train)


```

```python
y_pred = model.predict(x_test)
model.score(x_test,y_test)
```

```python
from sklearn.metrics import ConfusionMatrixDisplay as cmd

sns.set_theme()
fig,ax = plt.subplots(figsize=(8,6))
ax.grid(False)
cmd.from_estimator(model,x_test,y_test,cmap='Blues',colorbar=False ,ax =ax)
```
