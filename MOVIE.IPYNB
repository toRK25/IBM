```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

df = pd.read_csv('movies.csv')
df = df.head(5000)  # Limit to first 5000 rows
df.head()
```

```python
df = df[['title', 'genres', 'movieId']]
df =df.fillna('')
df.head()
df.info()
```

```python
# Remove year from title (e.g., "Toy Story (1995)" -> "Toy Story")
df['title'] = df['title'].str.replace(r'\s*\(\d{4}\)', '', regex=True)
df.head()
```

```python
df['features'] = df['title'] + ' ' + df['genres'] + ' ' + df['movieId'].astype(str)
```

```python
import sklearn
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(stop_words='english',min_df=20)
word_matrix = cv.fit_transform(df['features'])
word_matrix.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
sim = cosine_similarity(word_matrix)
```

```python
def recommend(title,df,sim,count=10):
    index = df.index[df['title'].str.lower() == title.lower()]

    if len(index) == 0:
        return "Movie not found in the database."
    
    similar_movies = list(enumerate(sim[index[0]]))
    recommendation = sorted(similar_movies, key=lambda x: x[1], reverse=True)

    top_recs = recommendation[1:count+1]
    titles = []
    for i in range(len(top_recs)):
        title = df.iloc[top_recs[i][0]]['title']
        titles.append(title)
    return titles
```

```python
recommend('Toy Story (1995)',df,sim)
```
